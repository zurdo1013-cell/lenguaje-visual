## CALCULADORA FUNCIONAL
# realiza una  calculadora funcional en python con las 4 operaciones basicas.
el programa debe tener un menu interactivo y utiliza funciones con las 4 operaciones
debe tener todas las validaciones que consideres necesaria

```python

print("CALCULADORA")

while True:

    print("""
    ¿Qué quieres hacer? Escribe una opción:

    1) Restar
    2) Sumar
    3) Multiplicar
    4) Dividir
    5) Salir
    """)

    opcion = input("Opción: ")

    if opcion == '1':

        n1 = float(input("Ingrese número: "))
        n2 = float(input("Ingrese número: "))

        print("El resultado de la resta es:", n1 - n2)

    elif opcion == '2':

        n1 = float(input("Ingrese número: "))
        n2 = float(input("Ingrese número: "))

        print("El resultado de la suma es:", n1 + n2)

    elif opcion == '3':

        n1 = float(input("Ingrese número: "))
        n2 = float(input("Ingrese número: "))

        print("El resultado de la multiplicación es:", n1 * n2)

    elif opcion == '4':

        n1 = float(input("Ingrese número: "))
        n2 = float(input("Ingrese número: "))

        if n2 != 0:
            print("El resultado de la división es:", n1 / n2)
        else:
            print("No puedes dividir por 0")

    elif opcion == '5':

        print("Hasta luego")
        break

    else:

        print("ERROR EN EL SISTEMA")
```

sirve para hcaer referencia a metodos y atributos base de una clase dentro de sus propios
metodos

```python
class galleta():
    chocolate = False
    def __init__(self):
        print("se acaba de crear una galleta.")
        
    def chocolatear(self):
        self.chocolate = True
    def tiene_chocolate(self):
        if(self.chocolate):
            print("soy una galleta chocolateada:>_<")
        else:
            print("soy una galleta sin chocolate :-(")
```

# #investigar un metodo especiales de clase
# contructor :
El constructor es un método especial de una clase que se ejecuta automáticamente cuando se crea un objeto.__init__()
# destructor :
 El destructor es un método especial que se ejecuta cuando un objeto está siendo eliminado de la memoria. __del__()

## Diferencia entre constructor y destructor
# Constructor	Destructor
__init__()	__del__()
Se ejecuta al crear el objeto	Se ejecuta al destruir/eliminar el objeto
Inicializa atributos	Realiza tareas de limpieza
Prepara el objeto para utilizarlo	Finaliza el uso del objeto
Resumen para tu investigación

# Constructor: 
método especial utilizado para inicializar un objeto cuando este es creado.

# Destructor: 
método especial utilizado para realizar tareas de limpieza cuando un objeto deja de existir.

# Otros métodos especiales de Python son, por ejemplo: 
__str__(), __len__(), __add__() y __eq__(). 
Estos permiten definir cómo se comportan los objetos ante determinadas operaciones.

```python
# constructor
class Persona:

    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad


persona1 = Persona("jose", 18)

print(persona1.nombre)
print(persona1.edad)
```

```python
# destructor
class Persona:

    def __init__(self, nombre):
        self.nombre = nombre
        print("Objeto creado")

    def __del__(self):
        print("Objeto eliminado")


persona1 = Persona("Isaias")

del persona1
```
