# slicing en la cadena
es una capacidad de las cadenas que devuelven un subconjunto o subcadena utilizando 2 indices[inicio:fin]
-el primer indice indica donde empieza la subcadena('se incluye el caracter)
-el segundo indice indica donde acaba la subcadena("se excluye el caracter")

```python
palabra ="python"
```

```python
palabra[2:6]
```

```python
palabra[:2]
```

```python
palabra[2:]
```

```python
palabra[:]
```

```python
palabra[:2] + palabra[2:]
```

```python
palabra[-2:]
```

```python
palabra[99]
```

```python
palabra[:99]
```

```python
palabra[99:]
```
