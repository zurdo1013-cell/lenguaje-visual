## Ejercicio 3 — Calculadora de suma

Crea dos campos numéricos y un botón **Sumar**. Muestra el resultado en una etiqueta. Si algún dato no es numérico, muestra un error.

**Pista:** utiliza `try` y `except ValueError`.

**Conceptos:** `Entry`, conversión de datos, validación y `messagebox`.

```python
#ejercicio 3 
import tkinter as tk
from tkinter import messagebox

def sumar():
    try:
        numero1 = int(entrada1.get())
        numero2 = int(entrada2.get())

        resultado = numero1 + numero2

        etiqueta_resultado.config(text=f"Resultado: {resultado}")

    except ValueError:
        messagebox.showerror(
            "Error",
            "Por favor, ingresa solamente números."
        )

ventana = tk.Tk()
ventana.title("Calculadora de suma")
ventana.geometry("300x300")


titulo = tk.Label(
    ventana,
    text="Calculadora de suma",
    font=("Arial", 18),
    fg="green"
)
titulo.pack(pady=20)

entrada1 = tk.Entry(ventana)
entrada1.pack(pady=5)

entrada2 = tk.Entry(ventana)
entrada2.pack(pady=5)

boton_sumar = tk.Button(
    ventana,
    text="Sumar",
    command=sumar,
    bg= "#53920b"
)
boton_sumar.pack(pady=15)


etiqueta_resultado = tk.Label(
    ventana,
    text="Resultado: ",
    font=("Arial", 14),
    fg="red"
)
etiqueta_resultado.pack(pady=10)

ventana.mainloop()
```

## Ejercicio 4 — Conversor de guaraníes a dólares

Solicita un monto en guaraníes y una cotización del dólar. Al presionar **Convertir**, calcula y muestra el equivalente en dólares con dos decimales.

Valida que:

- Ningún campo esté vacío.
- Los valores sean numéricos y mayores que cero.
- La cotización no sea cero.

**Conceptos:** formulario con `grid()`, cálculos y validación.

```python

import tkinter as tk

def convertir():
    monto = entrada_monto.get()
    cotizacion = entrada_cotizacion.get()

    resultado.config(text="")

    if monto == "" or cotizacion == "":
        resultado.config(text="Error: ningún campo puede estar vacío.")
        return

    try:
        monto = float(monto)
        cotizacion = float(cotizacion)

        
        if monto <= 0:
            resultado.config(text="Error: el monto debe ser mayor que cero.")
            return

        if cotizacion <= 0:
            resultado.config(text="Error: la cotización debe ser mayor que cero.")
            return

        dolares = monto / cotizacion

        resultado.config(text=f"Resultado: ${dolares:.2f}")

    except ValueError:
        resultado.config(text="Error: ingresa solamente números.")

ventana = tk.Tk()
ventana.title("Conversor de Guaraníes a Dólares")
ventana.geometry("450x280")


titulo = tk.Label(
    ventana,
    text="Conversor de Gs. a USD",
    font=("Arial", 18),
    fg= "blue"
)
titulo.grid(row=0, column=0, columnspan=2, pady=20)

tk.Label(
    ventana,
    text="Monto en guaraníes:",
    fg="#148a04"
).grid(row=1, column=0, padx=10, pady=10)

entrada_monto = tk.Entry(ventana)
entrada_monto.grid(row=1, column=1, padx=10, pady=10)

tk.Label(
    ventana,
    text="Cotización del dólar:",
    fg="green"
).grid(row=2, column=0, padx=10, pady=10)

entrada_cotizacion = tk.Entry(ventana)
entrada_cotizacion.grid(row=2, column=1, padx=30, pady=10)

boton_convertir = tk.Button(
    ventana,
    text="Convertir",
    command=convertir,
    fg="red"
)
boton_convertir.grid(row=3, column=0, columnspan=2, pady=15)

resultado = tk.Label(
    ventana,
    text="resultado:",
    font=("Arial", 12),
   bg="#33c70e"
)
resultado.grid(row=4, column=0, columnspan=2, pady=10)

ventana.mainloop()
```

# Ejercicio 5 — Selección de curso

Crea un formulario con:

- Campo para el nombre.
- `Radiobutton` para elegir Primero, Segundo o Tercer año.
- `Checkbutton` para indicar si el estudiante tiene computadora.
- Botón **Mostrar resumen**.

Presenta toda la información en una etiqueta.

**Conceptos:** `StringVar`, `BooleanVar`, `Radiobutton` y `Checkbutton`.

```python

import tkinter as tk

ventana = tk.Tk()
ventana.title("Selección de curso")
ventana.geometry("400x350")


nombre = tk.StringVar()
Año = tk.StringVar(value="Primero")
tiene_computadora = tk.BooleanVar(value=False)



def mostrar_resumen():
    nombre_estudiante = nombre.get()
    curso = Año.get()
    computadora = tiene_computadora.get()

    if computadora:
        computadora_texto = "Sí"
    else:
        computadora_texto = "No"

    resumen.config(
        text=f"Nombre: {nombre_estudiante}\n"
             f"Año: {curso}\n"
             f"Tiene computadora: {computadora_texto}"
    )


titulo = tk.Label(
    ventana,
    text="Selección de curso",
    font=("Arial", 18),fg="orange"
)
titulo.grid(row=0, column=0, columnspan=2, pady=20)



tk.Label(
    ventana,
    text="Nombre:",bg="purple"
).grid(row=1, column=0, padx=10, pady=10, sticky="e")

entrada_nombre = tk.Entry(
    ventana,
    textvariable=nombre
)
entrada_nombre.grid(row=1, column=1, padx=10, pady=10)


tk.Label(
    ventana,
    text="Año:",bg="purple"
).grid(row=2, column=0, padx=10, pady=10, sticky="e")


tk.Radiobutton(
    ventana,
    text="Primero",
    variable=Año,
    value="Primero",fg="red"
).grid(row=2, column=1, sticky="w")

tk.Radiobutton(
    ventana,
    text="Segundo",
    variable=Año,
    value="Segundo",fg="red"
).grid(row=3, column=1, sticky="w")

tk.Radiobutton(
    ventana,
    text="Tercer",
    variable=Año,
    value="Tercer",fg="red"
).grid(row=4, column=1, sticky="w")


computadora = tk.Checkbutton(
    ventana,
    text="Tengo computadora",
    variable=tiene_computadora,fg="green"
)
computadora.grid(row=5, column=1, sticky="w", pady=10)


boton = tk.Button(
    ventana,
    text="Mostrar resumen",
    command=mostrar_resumen,bg="orange"
)
boton.grid(row=6, column=0, columnspan=2, pady=15)


resumen = tk.Label(
    ventana,
    text="",
    font=("Arial", 12),fg="blue"
  
)
resumen.grid(row=7, column=0, columnspan=2, pady=10)

ventana.mainloop()

```

## Ejercicio 6 — Editor de nota breve

- **Mostrar:** presenta el contenido en un `messagebox`.
- **Limpiar:** elimina todo el texto.
- **Salir:** solicita confirmación antes de cerrar.

**Conceptos:** `Text`, cuadros de diálogo y eventos.

## Ejercicio 7 — Inicio de sesión simulado

Crea campos para usuario y contraseña. La contraseña debe mostrarse con asteriscos usando `show="*"`.

Reglas:

- Usuario correcto: `admin`.
- Contraseña correcta: `1234`.
- Si ambos coinciden, mostrar acceso concedido.
- En caso contrario, mostrar acceso denegado.
- Permitir iniciar sesión presionando Enter.

**Conceptos:** `Entry`, condicionales, `messagebox` y `bind()`.

```python

import tkinter as tk
from tkinter import messagebox



def iniciar_sesion(event=None):
    usuario = entrada_usuario.get()
    contraseña = entrada_contraseña.get()

    if usuario == "admin" and contraseña == "1234":
        messagebox.showinfo(
            "Acceso",
            "Acceso concedido"
        )
    else:
        messagebox.showerror(
            "Acceso",
            "Acceso denegado"
        )



ventana = tk.Tk()
ventana.title("Inicio de sesión")
ventana.geometry("350x250")



titulo = tk.Label(
    ventana,
    text="Inicio de sesión",
    font=("Arial", 18)
)
titulo.pack(pady=20)



tk.Label(
    ventana,
    text="Usuario:",fg="red"
).pack()

entrada_usuario = tk.Entry(ventana)
entrada_usuario.pack(pady=5)



tk.Label(
    ventana,
    text="Contraseña:",fg="red"
).pack()

entrada_contraseña = tk.Entry(
    ventana,
    show="*"
)
entrada_contraseña.pack(pady=5)


boton = tk.Button(
    ventana,
    text="Iniciar sesión",
    command=iniciar_sesion,fg="red"
)
boton.pack(pady=15)


ventana.bind("<Return>", iniciar_sesion)

ventana.mainloop()

```

## Ejercicio 8 — Proyecto integrador: registro de visitantes

Construye la aplicación propuesta al inicio de la clase.

### Requisitos mínimos

1. Título y tamaño apropiado.
2. Encabezado con el nombre del sistema.
3. Campos: nombre, documento y motivo de visita.
4. Botones: Registrar, Limpiar y Salir.
5. Validación de campos obligatorios.
6. Confirmación mediante `messagebox`.
7. Limpieza automática después de registrar.
8. Cursor colocado nuevamente en el campo nombre.
9. Tecla Enter asociada a la acción Registrar.
10. Organización mediante `Frame` y `grid()`.

### Desafío adicional

Guarda temporalmente cada registro en una lista y muestra en una etiqueta cuántos visitantes fueron registrados.

```python

import tkinter as tk
from tkinter import messagebox


visitantes = []

def registrar():
    nombre = entrada_nombre.get()
    documento = entrada_documento.get()
    motivo = entrada_motivo.get()


    if nombre == "" or documento == "" or motivo == "":
        messagebox.showerror(
            "Error",
            "Todos los campos son obligatorios."
        )
        return

   
    visitante = {
        "nombre": nombre,
        "documento": documento,
        "motivo": motivo
    }

    visitantes.append(visitante)

    messagebox.showinfo(
        "Registro",
        "Visitante registrado correctamente."
    )

   
    cantidad.config(
        text=f"Visitantes registrados: {len(visitantes)}"
    )

    limpiar()

    entrada_nombre.focus()


def limpiar():
    entrada_nombre.delete(0, tk.END)
    entrada_documento.delete(0, tk.END)
    entrada_motivo.delete(0, tk.END)


def salir():
    respuesta = messagebox.askyesno(
        "Salir",
        "¿Deseas salir del sistema?"
    )

    if respuesta:
        ventana.destroy()


ventana = tk.Tk()
ventana.title("Registro de Visitantes")
ventana.geometry("500x400")


frame_titulo = tk.Frame(ventana)
frame_titulo.pack(pady=20)

titulo = tk.Label(
    frame_titulo,
    text="SISTEMA DE REGISTRO DE VISITANTES",
    font=("Arial", 18)
)
titulo.grid(row=0, column=0)


frame_formulario = tk.Frame(ventana)
frame_formulario.pack(pady=10)


tk.Label(
    frame_formulario,
    text="Nombre:"
).grid(row=0, column=0, padx=10, pady=10, sticky="e")

entrada_nombre = tk.Entry(
    frame_formulario,
    width=30
)
entrada_nombre.grid(row=0, column=1, padx=10, pady=10)


tk.Label(
    frame_formulario,
    text="Documento:"
).grid(row=1, column=0, padx=10, pady=10, sticky="e")

entrada_documento = tk.Entry(
    frame_formulario,
    width=30
)
entrada_documento.grid(row=1, column=1, padx=10, pady=10)

tk.Label(
    frame_formulario,
    text="Motivo de visita:"
).grid(row=2, column=0, padx=10, pady=10, sticky="e")

entrada_motivo = tk.Entry(
    frame_formulario,
    width=30
)
entrada_motivo.grid(row=2, column=1, padx=10, pady=10)


frame_botones = tk.Frame(ventana)
frame_botones.pack(pady=15)


boton_registrar = tk.Button(
    frame_botones,
    text="Registrar",
    command=registrar
)
boton_registrar.grid(row=0, column=0, padx=5)


boton_limpiar = tk.Button(
    frame_botones,
    text="Limpiar",
    command=limpiar
)
boton_limpiar.grid(row=0, column=1, padx=5)


boton_salir = tk.Button(
    frame_botones,
    text="Salir",
    command=salir
)
boton_salir.grid(row=0, column=2, padx=5)


cantidad = tk.Label(
    ventana,
    text="Visitantes registrados: 0",
    font=("Arial", 12)
)
cantidad.pack(pady=15)

ventana.bind("<Return>", lambda event: registrar())

entrada_nombre.focus()

ventana.mainloop()

```
