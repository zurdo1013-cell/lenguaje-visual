# funciones recursiva
se trata de funciones que se llaman a si mismo durante su propia ejecucion, funciones de forma similar a lasa iteraciones y debemos encargarnos de palnificar el momento en que una funcion recursiva deje de llamarse 0 tendremos una funcion recursiva infinita.

suele utilizarse para dividir una tarea o sbtarea mas simple de forma qu sea facil  de abordar el problema y solucionarlo.

```python
# ejemplo contador hasta 0
def cuenta_atras(numero):
    numero -=1
    if numero >0:
        print(numero)
        cuenta_atras(numero)
    else:
        print("chakeeeeeee")
    print("fin de la funcion",numero)
    
```

```python
cuenta_atras(5)
```

```python
# realiza una funcion llamada area_circulo() que devuelva el area del circulo a partir de  un radio. 
# calcular el area de un circulo de 5 de radio. el area del circulo se obtiene al elevar el radio a dor y multiplicamos
# por el numero pi. puedes utilizar el valor de 3,14159 como pi.
def area_circulo(radio):
    return (radio**2)* pi
pi= 3.14159
area = area_circulo(5)
print("el area del circuloo es:",area)
```

```python
#realiza una funcion relacion() que a partir de dos numeros cumpla los sgtes:
# si el numero es mayor que el segundo, debe devolver 1.
# si el primer numero es menor que el segundo,debe devolver -1.
# si ambos son iguales, debe devolver 0.
#comprueba la relacion entre los numeros: 5 y 10, 10y5, 5y5

def relacion(a, b):
    if a > b:
        return 1
    elif a < b:
        return -1
    else:
        return 0
print(relacion(5, 10))   
print(relacion(10, 5))   
print(relacion(5, 5))   
```

```python
# resliza una funcion llamada intermedio()que a partir de dos numeros, devuelve su punto intermedio:
# nota: el numero intermedio de dos numeros corresponde a la suma de 2 numeros divida entre 2
# comprueba el punto intermedio entre -12y24.
def intermedio(a, b):
    return sum([a, b]) / 2

print(intermedio(-12, 24))
```
