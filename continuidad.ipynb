## iteracion
iteracion signiifica realizar una accion varias veces. cada vez que se repite se denomina iteracion.

## sentencia while (mientras)
se basa en repetir un bloque a partir de evaluar una condicion logico siempre que esta sea true
queda en las manos del programador decidir el momento en que la condicion cambie a false para hacer que el while finalice


```python
c = 0
while c <= 5:
    c+=1
    print("c vale",c)
```

## sentencia else en bucle while
se encadena al while para ejecutar un bloque de codigo una vez la condicion ya no devuelva true(normalmente al final)

```python
c = 0
while c<=5:
    c+=1
    print("c vale",c)
else:
    print("se ha completado toda la iteracion y c vale",c)
    
```

## interacion break
sirve para romper la ejecucion del while en cualquier mmento no se ejecutara el else ya que este solo se llama al final la iteracion

```python
c = 0
while c<=5:
    c+=1
    if(c==4):
        print("rompemos el bucle cuando c vale",c)
        break
    print("c vale",c)
else:
    print("se ha completado toda la iteracion y c vale",c)
    
```

## iteracion continue..
sirve para saltarse las iteraciones sin romper el bucle 

```python
c = 0
while c<=5:
    c+=1
    if c==3 or c==4:
        #print("continuamos con la iteracion",c)
        continue
    print("c vale",c)
else:
    print("se ha completado toda la iteracion y c vale",c)
    
```

## creando un menu iteractivo


```python
print("bienvenidos al menu iteractiv0")
while (True):
    print(""" que quieres hacer? escribe una opcion
          1.saludar
          2.sumar dos numeros
          3.salir""")
    opcion = input()
    if opcion =='1':
        print("hola espero q este bien")
    elif opcion == '2':
        n1 = float(input("ingrese numero"))
        n2 =  float(input("ingrese numero"))
        print("el resultado de la suma es:",n1+n2)
    elif opcion == '3':
        print("hasta luego")
        break
    else:
        print("comando desconocido vuelva a intentar")
        
```

## ejercicios
1) realiza un programa que lea 2 numeros por teclado y permita elegir entre 3 opciones en un menu:
- mostrar una suma de dos numeros
- mostrar una resta de 2 numeros( primero menos elsegundo)
- mostrar multiplicacion
- en caso de no introducir una opcion valida, el programa informara  de que no es correcta.

```python
print("BIENVENIDO AL MENU")
while (True):
    print(""" que quieres hacer? escribe una opcion
          1) restar dos numeros
          2) sumar dos numeros
          3) multiplicar dos numeros
          4) salir""")
    opcion = input()
    if opcion =='1':
        n1 = float(input("ingrese numero"))
        n2 =  float(input("ingrese numero"))
        print("el resultado de la suma es:",n1-n2)
    elif opcion == '2':
        n1 = float(input("ingrese numero"))
        n2 =  float(input("ingrese numero"))
        print("el resultado de la suma es:",n1+n2)

    elif opcion == '3':
       n1 = float(input("ingrese numero"))
       n2 =  float(input("ingrese numero"))
       print("el resultado de la suma es:",n1*n2)
       
    elif opcion == '4':
     print("hasta luego")
     break
   
    else:
        print("ERRRO EN EL SYSTEM")
```

4) realiza un programa que pida al usuario cuantos numeros quiere introducir. luego lee todo lo numeros y realiza una media aritmetica

```python
n = int(input("¿Cuántos números quieres introducir? "))

suma = 0
contador = 0
while contador < n:
    numero = int(input("Introduce un número: "))
    suma = suma + numero
    contador = contador + 1
    media = suma / n

print("La media aritmética es:", media)
```
