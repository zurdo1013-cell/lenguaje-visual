```python
import tkinter as tk
from tkinter import messagebox
import re

# ==========================================================
# RADAR DE OPERADOR - V1
# Paraguay
# ==========================================================

NOMBRE_APP = "RADAR DE OPERADOR V1"


def limpiar_numero(numero):
    numero = numero.strip()

    # Dejar solamente números
    numero = re.sub(r"\D", "", numero)

    # Si viene con +595
    if numero.startswith("595"):
        numero = numero[3:]

    # Si viene como 0 + número
    if numero.startswith("0"):
        numero = numero[1:]

    return numero


def validar_numero(numero):
    numero = limpiar_numero(numero)


    if len(numero) != 9:
        return False

    # Debe comenzar con 9
    if not numero.startswith("9"):
        return False

    return True

def consultar_operador(numero):
    numero = limpiar_numero(numero)

    prefijo = numero[:4]
    prefijos = {
    
        "0971": "personal",
        "0972": "personal",
        "0973": "personal",
        "0974": "personal",

        "0981": "Tigo",
        "0982": "Tigo",
        "0983": "Tigo",
        "0984": "Tigo",

        "0991": "claro",
        "0992": "claro",
        "0993": "claro",
        "0994": "claro",

    }

    operador_referencial = prefijos.get(prefijo)

    if operador_referencial:
        return { "numero": "0" + numero,"operador": operador_referencial,"tipo": "REFERENCIAL",
            "estado": "OK"
        }
    return {   "numero": "0" + numero, "operador": "No determinado", "tipo": "SIN DATOS",
        "estado": "OK"
    }


# ==========================================================
# BOTÓN CONSULTAR
# ==========================================================

def consultar():
    numero = entrada_numero.get()

    if not validar_numero(numero):
        messagebox.showerror(
            "Número inválido",
            "Ingrese número  válido.\n\n"
            "Ejemplo:\n"
            "098","099","097"
        )
        return

    resultado = consultar_operador(numero)

    # Limpiar resultado
    etiqueta_numero.config(
        text=f"NÚMERO\n{resultado['numero']}"
    )

    etiqueta_operador.config(
        text=f"OPERADOR\n{resultado['operador']}"
    )

    etiqueta_tipo.config(
        text=f"TIPO DE RESULTADO\n{resultado['tipo']}"
    )

    etiqueta_estado.config(
        text=f"ESTADO\n{resultado['estado']}"
    )


# ==========================================================
# LIMPIAR
# ==========================================================

def limpiar():
    entrada_numero.delete(0, tk.END)

    etiqueta_numero.config(
        text="NÚMERO\n---"
    )

    etiqueta_operador.config(
        text="OPERADOR\n---"
    )

    etiqueta_tipo.config(
        text="TIPO DE RESULTADO\n---"
    )

    etiqueta_estado.config(
        text="ESTADO\n---"
    )


# ==========================================================
# VENTANA
# ==========================================================

ventana = tk.Tk()

ventana.title(NOMBRE_APP)
ventana.geometry("600x520")
ventana.resizable(False, False)

# Fondo
ventana.configure(bg="#031631")


# ==========================================================
# TÍTULO
# ==========================================================

titulo = tk.Label(
    ventana,
    text="📡 RADAR DE OPERADOR",
    font=("Segoe UI", 24, "bold"),
    fg="#db0a0a",
    bg="#07111f"
)

titulo.pack(pady=(25, 5))


subtitulo = tk.Label(
    ventana,
    text="Consulta de numeración móvil - Paraguay",
    font=("Segoe UI", 11),
    fg="#769ebd",
    bg="#07111f"
)

subtitulo.pack(pady=(0, 25))


# ==========================================================
# ENTRADA
# ==========================================================

marco_entrada = tk.Frame(
    ventana,
    bg="#0d1b2a",
    padx=20,
    pady=20
)

marco_entrada.pack(
    padx=35,
    fill="x"
)


texto = tk.Label(
    marco_entrada,
    text="Ingrese el número:",
    font=("Segoe UI", 11, "bold"),
    fg="white",
    bg="#0d1b2a"
)

texto.pack(anchor="w")


entrada_numero = tk.Entry(
    marco_entrada,
    font=("Segoe UI", 16),
    justify="center",
    bg="#111f30",
    fg="#00eaff",
    insertbackground="#00eaff",
    relief="flat"
)

entrada_numero.pack(
    pady=12,
    ipady=8,
    fill="x"
)

entrada_numero.insert(
    0,
    "0981234567"
)


marco_botones = tk.Frame(
    ventana,
    bg="#07111f"
)

marco_botones.pack(pady=20)


boton_consultar = tk.Button(
    marco_botones,
    text="🔎 CONSULTAR",
    font=("Segoe UI", 11, "bold"),
    bg="#007c91",
    fg="white",
    activebackground="#00a8c6",
    activeforeground="white",
    relief="flat",
    padx=25,
    pady=10,
    command=consultar
)

boton_consultar.grid(
    row=0,
    column=0,
    padx=8
)


boton_limpiar = tk.Button(
    marco_botones,
    text="LIMPIAR",
    font=("Segoe UI", 11, "bold"),
    bg="#26384a",
    fg="white",
    activebackground="#344d65",
    activeforeground="white",
    relief="flat",
    padx=25,
    pady=10,
    command=limpiar
)

boton_limpiar.grid(
    row=0,
    column=1,
    padx=8
)


marco_resultado = tk.Frame(
    ventana,
    bg="#0d1b2a",
    padx=20,
    pady=15
)

marco_resultado.pack(
    padx=35,
    fill="x"
)


etiqueta_numero = tk.Label(
    marco_resultado,
    text="NÚMERO\n---",
    font=("Consolas", 12, "bold"),
    fg="#00eaff",
    bg="#0d1b2a",
    justify="left"
)

etiqueta_numero.pack(
    anchor="w",
    pady=5
)


etiqueta_operador = tk.Label(
    marco_resultado,
    text="OPERADOR\n---",
    font=("Consolas", 12, "bold"),
    fg="#00ff9d",
    bg="#0d1b2a",
    justify="left"
)

etiqueta_operador.pack(
    anchor="w",
    pady=5
)


etiqueta_tipo = tk.Label(
    marco_resultado,
    text="TIPO DE RESULTADO\n---",
    font=("Consolas", 11),
    fg="#e2a01c",
    bg="#0d1b2a",
    justify="left"
)

etiqueta_tipo.pack(
    anchor="w",
    pady=5
)


etiqueta_estado = tk.Label(
    marco_resultado,
    text="ESTADO\n---",
    font=("Consolas", 11),
    fg="#8ca8bd",
    bg="#0d1b2a",
    justify="left"
)

etiqueta_estado.pack(
    anchor="w",
    pady=5
)


# ==========================================================
# ADVERTENCIA
# ==========================================================

advertencia = tk.Label(
    ventana,
    text=(
        "⚠ La información por prefijo es referencial.\n"
        "La portabilidad numérica puede cambiar el operador actual."
    ),
    font=("Segoe UI", 9),
    fg="#8295a8",
    bg="#07111f",
    justify="center"
)

advertencia.pack(
    pady=15
)


# ==========================================================
# INICIAR
# ==========================================================

ventana.mainloop()
```
